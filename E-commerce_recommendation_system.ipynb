# Rating-Based Recommendation System

```python
df['product_rating'] = pd.to_numeric(df['product_rating'], errors='coerce')
df = df.dropna(subset=['product_rating'])  

product_rating_df = df.groupby('uniq_id').agg({'product_rating': ['mean', 'count'],'product_name': 'first'}).reset_index()

product_rating_df.columns = ['ProdID', 'AvgRating', 'RatingCount', 
                             'ProductName']
```

```python
def recommend_top_products(n=10, min_ratings=1):
    popular = product_rating_df[product_rating_df['RatingCount'] >= min_ratings]
    popular = popular.sort_values(by='AvgRating', ascending=False).head(n)
    return popular[['ProductName', 'AvgRating', 'RatingCount']]

top_products = recommend_top_products(n=10, min_ratings=1)
print(top_products)
```

#  Content-Based Recommendation System


```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python

df['combined_features'] = (df['product_category_tree'] + ' ' +df['brand'] + ' ' +df['product_specifications'] + ' ' +df['description'] + ' ' +df['product_name'])
```

```python
tfidf = TfidfVectorizer(stop_words='english')
tfidf_matrix = tfidf.fit_transform(df['combined_features'])
print("TF-IDF matrix shape:", tfidf_matrix.shape)
```

```python
cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)
```

```python
df = df.reset_index()

indices = pd.Series(df.index, index=df['product_name']).drop_duplicates()
```

```python
def content_based_recommendations(product_name, top_n=5):
    idx = indices.get(product_name)
    if idx is None:
        return "Product not found!"

    # Get similarity scores
    sim_scores = list(enumerate(cosine_sim[idx]))

    # Sort scores
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)

    # Top-N similar (excluding itself)
    sim_scores = sim_scores[1:top_n+1]

    # Get indices
    product_indices = [i[0] for i in sim_scores]

    # Return recommended products
    return df[['product_name', 'overall_rating', 'brand', 'image', 'product_rating']].iloc[product_indices]
```

# Collaborative Filtering System (User-Based)


```python
ratings_df = df[['uniq_id', 'pid', 'product_rating']].copy()

# Rename for clarity
ratings_df.columns = ['UserID', 'ProductID', 'Rating']


```

```python
# Create user-item matrix
user_item_matrix = ratings_df.pivot_table(index='UserID', columns='ProductID', values='Rating')

# Fill missing ratings (0 = no interaction)
user_item_matrix_filled = user_item_matrix.fillna(0)
```

```python
user_item_matrix_filled
```

```python
# Compute cosine similarity between users
user_similarity = cosine_similarity(user_item_matrix_filled)

# Convert to DataFrame
user_similarity_df = pd.DataFrame(user_similarity,index=user_item_matrix_filled.index,columns=user_item_matrix_filled.index)
```

```python
user_similarity_df
```

```python
def recommend_products(user_id, top_n=5):
    if user_id not in user_similarity_df.index:
        return "User ID not found!"

    # Get similar users (excluding itself)
    similar_users = user_similarity_df[user_id].sort_values(ascending=False)[1:]

    weighted_ratings = pd.Series(dtype='float64')

    # Loop through similar users
    for sim_user, similarity_score in similar_users.items():
        sim_user_ratings = user_item_matrix.loc[sim_user]
        sim_user_ratings = sim_user_ratings[sim_user_ratings.notna()]

        # Weighted ratings
        weighted_ratings = weighted_ratings.add(sim_user_ratings * similarity_score, fill_value=0)

    # Normalize by similarity sum
    similarity_sums = user_similarity_df[user_id].drop(user_id).sum()
    final_scores = weighted_ratings / similarity_sums

    # Remove already rated products
    user_rated = user_item_matrix.loc[user_id][user_item_matrix.loc[user_id].notna()].index
    recommendations = final_scores.drop(index=user_rated, errors='ignore')

    # Get top product IDs
    top_product_ids = recommendations.sort_values(ascending=False).head(top_n).index

    # Merge with product details
    product_details = df[df['uniq_id'].isin(top_product_ids)][
        ['product_name', 'overall_rating', 'brand', 'image', 'product_rating']
    ].drop_duplicates()

    return product_details.head(top_n)
```

# Hybrid Recommendation System

```python
def hybrid_recommendations(user_id, product_name, top_n=5, alpha=0.5):
    # --- 1. Get content-based recommendations ---
    cb_df = content_based_recommendations(product_name, top_n=20)  # your existing function
    cb_df = cb_df.reset_index(drop=True)
    cb_df['Rank_CB'] = cb_df.index + 1
    cb_df['Score_CB'] = (len(cb_df) - cb_df['Rank_CB'] + 1) / len(cb_df)  # normalize 0-1

    # --- 2. Get collaborative recommendations ---
    cf_df = recommend_products(user_id, top_n=20)  # your existing function
    cf_df = cf_df.reset_index(drop=True)
    cf_df['Rank_CF'] = cf_df.index + 1
    cf_df['Score_CF'] = (len(cf_df) - cf_df['Rank_CF'] + 1) / len(cf_df)

    # --- 3. Merge on product name (can also merge on uniq_id if both return it) ---
    merged = pd.merge(cb_df, cf_df,
                      on='product_name',
                      how='outer',
                      suffixes=('_CB', '_CF'))

    # Fill NaN scores with 0
    merged['Score_CB'] = merged['Score_CB'].fillna(0)
    merged['Score_CF'] = merged['Score_CF'].fillna(0)

    # --- 4. Weighted score ---
    merged['HybridScore'] = alpha * merged['Score_CB'] + (1 - alpha) * merged['Score_CF']

    # --- 5. Sort & select columns ---
    merged = merged.sort_values('HybridScore', ascending=False)

    return merged[['product_name', 'overall_rating_CB', 'brand_CB', 'image_CB', 'product_rating_CB']].head(top_n)
```
